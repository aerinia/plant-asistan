```python
# Gerekli paketi kurun
!pip install -q pyngrok

from google.colab import userdata
from pyngrok import ngrok
import subprocess
import time

# 1. Eski tünelleri temizle
ngrok.kill()

# 2. Gizli anahtarlardan Ngrok token'ını al ve ayarla
try:
    NGROK_TOKEN = userdata.get('NGROK_TOKEN')
    ngrok.set_auth_token(NGROK_TOKEN)
except userdata.SecretNotFoundError:
    print("HATA: 'NGROK_TOKEN' adında bir gizli anahtar bulunamadı. Lütfen sol menüden ekleyin.")
    raise

# 3. Streamlit'i arka planda başlat
print("Arayüz hazırlanıyor...")
subprocess.Popen(["streamlit", "run", "app.py", "--server.enableCORS=false", "--server.enableXsrfProtection=false"])
time.sleep(3) # Streamlit'in ayağa kalkması için kısa bir bekleme

# 4. Ngrok tünelini aç (Port 8501)
public_url = ngrok.connect(8501)
print(f"Uygulamanız şurada yayında: {public_url}")
```
